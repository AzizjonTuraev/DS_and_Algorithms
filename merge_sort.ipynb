Merge Sort is a divide-and-conquer sorting algorithm. It divides the array into two halves, recursively sorts each half, and then merges the sorted halves back together. This approach helps achieve efficient sorting with a time complexity of O(n log n) in all cases (best, average, and worst).

How Merge Sort Works: <br>
1. Divide: Split the array into two halves.
2. Recursively sort each half of the array.
3. Merge the sorted halves together by comparing their elements one by one and placing them in the correct order.

```python
def merge_sort(arr):

    print(f"Steps:", arr)

    # Base case: If the array has only one element, it's already sorted
    if len(arr) <= 1:
        return arr
    
    # Step 1: Divide the array into two halves
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]
    print(left_half, right_half)
    
    # Step 2: Recursively sort both halves
    left_sorted = merge_sort(left_half)
    right_sorted = merge_sort(right_half)
    # Step 3: Merge the sorted halves
    return merge(left_sorted, right_sorted)


def merge(left, right):
    result = []
    i = j = 0
    
    # Merge the two sorted lists
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    # If there are remaining elements in left or right, add them to the result
    result.extend(left[i:])
    result.extend(right[j:])
    print(result)
    return result
```

```python
# Example usage
arr = [64, 25, 12, 22, 11]
sorted_arr = merge_sort(arr)
print("Sorted array:", sorted_arr)
```

Explanation of the Code:
1. Base case: If the array has one or fewer elements, it is already sorted, and we return it as is.
2. Divide step: The array is split into two halves (left_half and right_half).
3. Recursive sort: Each half is recursively sorted by calling merge_sort on them.
4. Merge step: The merge function combines two sorted arrays into a single sorted array. It compares elements from both halves and appends the smaller element to the result list. After one of the halves is exhausted, the remaining elements from the other half are added to the result.

Time Complexity: <br>
Best, Average, and Worst Case: O(n log n)
- Each split operation takes O(log n) time because we keep dividing the array in half.
- Merging each level of recursion requires O(n) time because we are processing all elements at each level.
- So, the overall time complexity is O(n log n).

Space Complexity: <br>
O(n): Merge Sort requires additional space to store the left and right halves during the merging process.

Stability: <br>
- Stable Sort: Merge Sort is stable, meaning that if two elements have equal values, their relative order will be preserved in the sorted array.

When to Use Merge Sort: <br>
- When stability is important: Merge Sort is stable, which is useful when the relative order of equal elements must be maintained.
- For large datasets: It guarantees O(n log n) performance, making it efficient for large datasets.
- External sorting: Merge Sort is often used for sorting large datasets that cannot fit entirely in memory (e.g., when sorting files on disk).

Advantages of Merge Sort: <br>
- Guaranteed O(n log n) time complexity.
- Stable sort, which is useful for some applications.
- It can be adapted for external sorting, where data doesn't fit in memory and needs to be sorted in chunks.

Disadvantages of Merge Sort: <br>
- Space complexity: Requires O(n) additional space for merging, so it uses more memory compared to some other sorting algorithms like Quick Sort and Bubble Sort (which are in-place).
- It is slower than Quick Sort in practice for small datasets, even though it has the same O(n log n) time complexity.
