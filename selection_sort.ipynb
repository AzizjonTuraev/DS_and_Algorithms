To implement the Selection Sort algorithm in Python, here's a step-by-step breakdown of how it works:

Find the minimum element in the unsorted portion of the array. <br>
Swap that minimum element with the element at the beginning of the unsorted portion. <br>
Move the boundary between the sorted and unsorted portions of the array one step forward. <br>
Repeat the process until the entire array is sorted. <br>

```python
def selection_sort(arr):
    n = len(arr)
    
    # Traverse through all array elements
    for i in range(n):
        # Find the minimum element in remaining unsorted array
        print(f"Steps:", arr)
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        
        # Swap the found minimum element with the first element of the unsorted part
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        print(arr)
```

```python
# Example usage
arr = [64, 25, 12, 22, 11]
print("Original array", arr)
selection_sort(arr)
print("Sorted array:", arr)
```

The outer loop (for i in range(n)) iterates over each element. <br>
The inner loop (for j in range(i+1, n)) finds the minimum element in the unsorted portion of the array. <br>
After finding the minimum element, we swap it with the element at index i (the start of the unsorted section). <br>

Time Complexity: <br>
Best, Average, and Worst Case Time Complexity: O(n²), since we have two nested loops.

Space Complexity: <br>
O(1), as it sorts the array in place without using extra space.
