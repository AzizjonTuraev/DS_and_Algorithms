Insertion Sort is a simple comparison-based sorting algorithm that builds the final sorted array one element at a time. It's much like sorting playing cards in your hands, where you take one card at a time and insert it into its correct position among the already sorted cards.

Here’s how it works step by step:

Steps of Insertion Sort: <br>
Start with the second element in the list. (The first element is already considered sorted). <br>
Compare this element with the element(s) before it. <br>
Shift all elements larger than the current element one position to the right to make space for the current element. <br>
Insert the current element into its correct position. <br>
Repeat the process for all elements in the list until the entire list is sorted. <br>

```python
def insertion_sort(arr):
    # Start from the second element (index 1) and compare it with previous elements
    print("Begin", arr)

    for i in range(1, len(arr)):
        key = arr[i]  # The element to be placed in the correct position
        j = i - 1  # Index of the previous element

        # Shift elements that are greater than key to one position ahead
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
            print("---", arr)
        
        # Place the key at the correct position
        arr[j + 1] = key
        print("-", arr)

    print("End", arr)

```

```python
# Example usage
arr = [5, 2, 9, 1, 5, 6]
insertion_sort(arr)
print("Sorted array:", arr)
```

Explanation: <br>
insertion_sort(arr): The function takes an array arr and sorts it using the Insertion Sort algorithm. <br>
The outer loop starts from the second element and considers it as the element to be inserted. <br>
The while loop shifts the elements that are greater than the current element (key) one position to the right. <br>
Finally, the key is inserted at the correct position. <br>

Time Complexity: <br>
- Best case: O(n) when the array is already sorted. (Only one comparison per element).
- Worst case: O(n²) when the array is in reverse order. (Each element is compared to all previously sorted elements).
- Average case: O(n²).
                
Space Complexity: <br>
- O(1) because it sorts the array in place, meaning it doesn’t require extra space beyond the input list.

When to use Insertion Sort: <br>
It’s efficient for small datasets or arrays that are already partially sorted. <br>
It is simple to implement and has low overhead for small arrays, making it practical for those cases. <br>
