Quick Sort is a divide-and-conquer algorithm that works by selecting a pivot element from the array and partitioning the other elements into two sub-arrays according to whether they are less than or greater than the pivot. The sub-arrays are then sorted recursively.

Key Steps: <br>
Pick a pivot element from the array. <br>
Partition the array around the pivot, placing the pivot in its correct position (all smaller elements on the left, all larger elements on the right). <br>
Recursively sort the sub-arrays to the left and right of the pivot. <br>

Key Concepts: <br>
Partitioning: This step rearranges the elements in the array so that all elements less than the pivot are on one side, and all elements greater than the pivot are on the other side. <br>
Recursion: After partitioning, you recursively apply Quick Sort to the two sub-arrays (left and right of the pivot). <br>

```python
from random import randrange

def qsort(array):
    print(f"Steps:", array)

    if len(array) <2:
        return array
    else:
        pivot = array.pop(randrange(len(array)))
        less = [x for x in array if x <= pivot]
        greater = [x for x in array if x > pivot]

        print(less, pivot, greater)
        return qsort(less) + [pivot] + qsort(greater)
```

```python
# Example usage
arr = [64, 25, 12, 22, 11]
print("Original array", arr)
sorted_arr = qsort(arr)
print("Sorted array:", sorted_arr)
```

1. Base Case: If the length of the array is 1 or less, it is already sorted, so we return it. <br>
2. Pivot Selection: We pick the middle element of the array as the pivot. <br>
3. Partitioning: We create three lists: <br>
- left: Elements smaller than the pivot. <br>
- right: Elements greater than the pivot. <br>
4. Recursive Call: We recursively apply Quick Sort on the left and right sub-arrays and then combine them with the middle array. <br>

Time Complexity: <br>
Best and Average Case: O(n log n) — when the pivot divides the array into two roughly equal parts. <br>
Worst Case: O(n²) — when the pivot selection is poor (e.g., always selecting the smallest or largest element). However, with random pivot selection, the worst-case can be minimized. <br>

Space Complexity: <br>
O(log n) for the recursive stack space in the average case, though it can degrade to O(n) in the worst case. <br>

When to Use Quick Sort: <br>
Average-Case Efficiency: Quick Sort is often faster in practice compared to other O(n log n) algorithms (like Merge Sort or Heap Sort), especially for large datasets. <br>
In-place Sorting: Quick Sort sorts in place, so it doesn't require additional space for storing the sorted list. <br>

Limitations: <br>
Worst-case performance: If the pivot selection is poor, Quick Sort can degrade to O(n²). However, this is mitigated by good pivot selection strategies like random pivot or median of three.
