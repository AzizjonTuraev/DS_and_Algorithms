Bubble Sort is one of the simplest sorting algorithms. The basic idea is to repeatedly compare adjacent elements in the list and swap them if they are in the wrong order (i.e., the first element is larger than the second). This process is repeated until the entire list is sorted. It gets its name because the largest unsorted element "bubbles" up to its correct position at the end of the list with each pass.

How Bubble Sort Works: <br>
1. Start at the beginning of the list. <br>
2. Compare adjacent elements. If the current element is greater than the next element, swap them. <br>
3. After each pass, the largest element in the unsorted part of the list "bubbles up" to the correct position at the end. <br>
4. Repeat the process for the unsorted portion of the list (ignoring the last sorted element). <br>
5. Stop when no swaps are made in a pass (this means the list is sorted). <br>

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        print(f"Steps:", arr)
        for j in range(n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                print(arr)
```

```python
# Example usage
arr = [64, 25, 12, 22, 11]
print("Original array", arr)
# arr = bubble_sort(arr)
bubble_sort(arr)
print("Sorted array:", arr)
```

```python
def selection_sort(arr):
    n = len(arr)
    
    # Traverse through all array elements
    for i in range(n):
        print(f"Steps:", arr)
        # Find the minimum element in remaining unsorted array
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        
        # Swap the found minimum element with the first element of the unsorted part
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        print(arr)
```

```python
# Example usage
arr = [64, 25, 12, 22, 11]
print("Original array", arr)
# arr = bubble_sort(arr)
selection_sort(arr)
print("Sorted array:", arr)
```

Code Explanation <br>
1. Outer loop: The outer loop (for i in range(n)) represents each pass through the array. With each pass, the largest unsorted element gets placed at the end. <br>
2. Inner loop: The inner loop (for j in range(0, n-i-1)) compares each pair of adjacent elements and swaps them if needed. <br>
3. Flag (swapped): This flag is used to check if any swaps were made during the pass. If no swaps were made, the array is already sorted, and we can stop early to save unnecessary passes. <br>

Time Complexity:
Best Case (already sorted array): O(n), because the algorithm only needs one pass through the array and can detect that no swaps are needed. <br>
Average and Worst Case: O(n²), because in the worst case, the algorithm will need to perform n passes, and each pass involves n comparisons.<br>

Space Complexity:
O(1), because Bubble Sort is an in-place sorting algorithm, meaning it doesn't require additional memory beyond the original array. <br>

When to Use Bubble Sort:
- Educational purposes: Bubble Sort is often used to introduce sorting algorithms due to its simplicity.
- Small datasets: It can be useful for small lists, where the overhead of more complex algorithms might not be worth it.
- Nearly sorted arrays: If the array is already sorted or nearly sorted, Bubble Sort may perform better than its worst-case time complexity suggests.

Limitations:
- Inefficiency for large datasets: For large arrays, Bubble Sort is very inefficient compared to other algorithms like Quick Sort or Merge Sort.
- Too many comparisons: Even if the array is already sorted, Bubble Sort still performs a lot of unnecessary comparisons unless you optimize it with a flag to detect swaps.
